```python
import pandas as pd
import os

# =====================================================
# KONFIGURASI
# =====================================================

file_excel = r"Report Term MTD - 2026-09-30.xlsx"

# Folder output
root_folder = os.path.join(os.getcwd(), "Termination Report")

# =====================================================
# BACA EXCEL
# =====================================================

print("Membaca file Excel...")

df = pd.read_excel(file_excel)

# Bersihkan nama kolom
df.columns = df.columns.str.strip()

print("\nDaftar kolom:")
for col in df.columns:
    print(f"- {col}")

# =====================================================
# BUAT FOLDER UTAMA
# =====================================================

all_folder = os.path.join(root_folder, "ALL")
branch_folder = os.path.join(root_folder, "BRANCH")
ga_folder = os.path.join(root_folder, "GA")

os.makedirs(all_folder, exist_ok=True)
os.makedirs(branch_folder, exist_ok=True)
os.makedirs(ga_folder, exist_ok=True)

print("\nFolder utama berhasil dibuat.")

# =====================================================
# BUAT FOLDER BRANCH
# BERDASARKAN TER NAME
# =====================================================

print("\nMembuat folder BRANCH...")

branch_list = (
    df["Ter Name"]
    .dropna()
    .astype(str)
    .str.strip()
    .unique()
)

for branch in sorted(branch_list):
    folder_path = os.path.join(branch_folder, branch)
    os.makedirs(folder_path, exist_ok=True)
    print(f"[BRANCH] {branch}")

# =====================================================
# BUAT FOLDER GA
# KHUSUS TER NAME = HGA
# BERDASARKAN DIV_NAME
# =====================================================

print("\nMembuat folder GA...")

ga_list = (
    df.loc[
        df["Ter Name"].astype(str).str.strip().str.upper() == "HGA",
        "Div_name"
    ]
    .dropna()
    .astype(str)
    .str.strip()
    .unique()
)

for ga in sorted(ga_list):
    folder_path = os.path.join(ga_folder, ga)
    os.makedirs(folder_path, exist_ok=True)
    print(f"[GA] {ga}")

print("\n===================================")
print("SELESAI MEMBUAT FOLDER")
print("===================================")
print(f"Lokasi : {root_folder}")
```

```python
import pandas as pd
import os

# =====================================================
# FILE EXCEL
# =====================================================

file_excel = r"C:\Users\yudhaiv\OneDrive - Manulife\Project\create folder\Report Term MTD - 2026-09-30.xlsx"

root_folder = r"C:\Users\yudhaiv\OneDrive - Manulife\Project\create folder\TST"

# =====================================================
# BACA EXCEL
# =====================================================

df = pd.read_excel(file_excel)

# Bersihkan nama kolom
df.columns = df.columns.astype(str).str.strip()

print("=" * 50)
print("DAFTAR KOLOM")
print("=" * 50)

for col in df.columns:
    print(repr(col))

# =====================================================
# FOLDER UTAMA
# =====================================================

root_folder = os.path.join(os.getcwd(), "TST")

all_folder = os.path.join(root_folder, "ALL")
branch_folder = os.path.join(root_folder, "BRANCH")
ga_folder = os.path.join(root_folder, "GA")

os.makedirs(all_folder, exist_ok=True)
os.makedirs(branch_folder, exist_ok=True)
os.makedirs(ga_folder, exist_ok=True)

# =====================================================
# BUAT FOLDER BRANCH
# =====================================================

print("\nMEMBUAT FOLDER BRANCH...")

branch_list = (
    df["Ter Name"]
    .fillna("")
    .astype(str)
    .str.strip()
)

branch_list = branch_list[branch_list != ""].unique()

for branch in sorted(branch_list):
    path = os.path.join(branch_folder, branch)
    os.makedirs(path, exist_ok=True)
    print(f"BRANCH : {branch}")

# =====================================================
# CEK ISI TER NAME
# =====================================================

print("\nISI TER NAME YANG DITEMUKAN")

ter_name_list = (
    df["Ter Name"]
    .fillna("")
    .astype(str)
    .str.strip()
    .unique()
)

for item in ter_name_list:
    print(repr(item))

# =====================================================
# BUAT FOLDER GA
# =====================================================

print("\nMEMBUAT FOLDER GA...")

# Ambil semua data HGA

hga_df = df[
    df["Ter Name"]
    .fillna("")
    .astype(str)
    .str.strip()
    .str.upper()
    == "HGA"
]

print(f"Jumlah baris HGA : {len(hga_df)}")

if len(hga_df) > 0:

    ga_list = (
        hga_df["Div_name"]
        .fillna("")
        .astype(str)
        .str.strip()
    )

    ga_list = ga_list[ga_list != ""].unique()

    print(f"Jumlah folder GA : {len(ga_list)}")

    for ga in sorted(ga_list):
        path = os.path.join(ga_folder, ga)
        os.makedirs(path, exist_ok=True)
        print(f"GA : {ga}")

else:
    print("TIDAK DITEMUKAN DATA HGA")

print("\nSELESAI")
```
