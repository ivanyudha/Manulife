```python
from PIL import Image
import pandas as pd
import shutil
import os

input_folder = r"C:\Users\yudhaiv\OneDrive - Manulife\JANICE DEBORA FREELY KUSUMAWATY's files - REQUEST KTP 1756 Agen (STTD OJK) - 10 Sep 2026\ID-Document"
output_folder = r"C:\Users\yudhaiv\OneDrive - Manulife\JANICE DEBORA FREELY KUSUMAWATY's files - REQUEST KTP 1756 Agen (STTD OJK) - 10 Sep 2026\ID-Document-PDF"
excel_file = r"C:\Users\yudhaiv\OneDrive - Manulife\JANICE DEBORA FREELY KUSUMAWATY's files - REQUEST KTP 1756 Agen (STTD OJK) - 10 Sep 2026\Rename KTP 1745 ag - 10 Sep 26.xlsx"

os.makedirs(output_folder, exist_ok=True)

df = pd.read_excel(excel_file, sheet_name="RENAME")

mapping = dict(zip(
    df["KODE AGEN"].astype(str).str.strip(),
    df["NAMED FILE"].astype(str).str.strip()
))

success_log = []
failed_log = []

for filename in os.listdir(input_folder):

    ext = os.path.splitext(filename)[1].lower()

    if ext not in [".jpg", ".jpeg", ".png", ".jfif", ".pdf"]:
        continue

    agent_code = filename.split("_")[0].strip()

    if agent_code not in mapping:

        failed_log.append({
            "FILE_ASLI": filename,
            "KODE_AGEN": agent_code,
            "ERROR": "Kode agen tidak ditemukan di Excel"
        })

        print(f"❌ Kode agen tidak ditemukan: {agent_code}")
        continue

    final_name = mapping[agent_code]

    source_path = os.path.join(input_folder, filename)
    pdf_path = os.path.join(output_folder, final_name + ".pdf")

    try:

        if ext == ".pdf":

            shutil.copy2(source_path, pdf_path)

        else:

            with Image.open(source_path) as img:
                img.convert("RGB").save(pdf_path)

        success_log.append({
            "FILE_ASLI": filename,
            "NAMA_BARU": final_name + ".pdf"
        })

        print(f"✅ {filename}")
        print(f"➡️ {final_name}.pdf")

    except Exception as e:

        failed_log.append({
            "FILE_ASLI": filename,
            "KODE_AGEN": agent_code,
            "ERROR": str(e)
        })

        print(f"❌ Gagal: {filename} | {e}")

pd.DataFrame(success_log).to_csv(
    os.path.join(output_folder, "LOG_BERHASIL.csv"),
    index=False,
    encoding="utf-8-sig"
)

pd.DataFrame(failed_log).to_csv(
    os.path.join(output_folder, "LOG_GAGAL.csv"),
    index=False,
    encoding="utf-8-sig"
)

print("\n==========================")
print(f"Berhasil : {len(success_log)}")
print(f"Gagal    : {len(failed_log)}")
print("==========================")
print("Log tersimpan:")
print("- LOG_BERHASIL.csv")
print("- LOG_GAGAL.csv")
```
