```python
from PIL import Image
import os

input_folder = r"C:\Users\yudhaiv\OneDrive - Manulife\ANDI ANDREA LESMANA's files - KTP PENGKINIAN 2025\E-Certificate Batch 2\KTP_HASIL_GABUNGAN"          
output_folder = r"C:\Users\yudhaiv\OneDrive - Manulife\ANDI ANDREA LESMANA's files - KTP PENGKINIAN 2025\E-Certificate Batch 2\KTP_HASIL_GABUNGAN\ktp_pdf_hasil"    

os.makedirs(output_folder, exist_ok=True)

success = 0
failed = 0

for filename in os.listdir(input_folder):
    if filename.lower().endswith((".jpg", ".jpeg", ".png")):
        img_path = os.path.join(input_folder, filename)
        pdf_name = os.path.splitext(filename)[0] + ".pdf"
        pdf_path = os.path.join(output_folder, pdf_name)
        
        try:
            with Image.open(img_path) as img:
                img = img.convert("RGB")
                img.save(pdf_path)
            success += 1
            print(f"✅ {filename} → {pdf_name}")
        except Exception as e:
            failed += 1
            print(f"❌ Gagal: {filename} | Error: {e}")
```
