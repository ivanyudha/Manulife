```python
import os
import pandas as pd
```

```python
folder_path = r"C:\Users\yudhaiv\OneDrive - Manulife\ANDI ANDREA LESMANA's files - KTP PENGKINIAN 2025\E-Certificate Batch 2\KTP_HASIL_GABUNGAN\ktp_pdf_hasil"
excel_path = r"C:\Users\yudhaiv\OneDrive - Manulife\ANDI ANDREA LESMANA's files - KTP PENGKINIAN 2025\E-Certificate Batch 2\Agen Code KTP 1700an.xlsx"
```

```python
df = pd.read_excel(excel_path)
```

```python
df.sample(5)
```

```python
if not {'OLD_NAME', 'NEW_NAME'}.issubset(df.columns):
    raise ValueError("Excel : OLD_NAME dan NEW_NAME")
```

```python
log_file = os.path.join(folder_path, "rename_log.txt")
with open(log_file, "w", encoding="utf-8") as log:
    for index, row in df.iterrows():
        old_key = str(row['OLD_NAME']).strip()
        new_name = str(row['NEW_NAME']).strip()

        found = False
        for file in os.listdir(folder_path):
            if old_key in file:
                old_path = os.path.join(folder_path, file)
                new_path = os.path.join(folder_path, new_name)

                if os.path.exists(new_path):
                    log.write(f"SKIPPED: {file} -> {new_name} (File Already)\n")
                    print(f"SKIPPED: {file} -> {new_name} (File Already)")
                else:
                    os.rename(old_path, new_path)
                    log.write(f"Renamed: {file} -> {new_name}\n")
                    print(f"Renamed: {file} -> {new_name}")
                found = True
                break

        if not found:
            log.write(f"NOT FOUND: OLD_NAME {old_key} tidak ditemukan\n")
            print(f"NOT FOUND: OLD_NAME {old_key} tidak ditemukan")

print(f"\nProses selesai. Log tersimpan di: {log_file}")
```
