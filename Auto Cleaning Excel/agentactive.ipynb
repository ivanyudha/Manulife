```python
import pandas as pd
nasd = pd.read_excel(r"C:\Users\yudhaiv\OneDrive - Manulife\Project\Contact Agent Active\NASD 2023 - 2026-10-05 pagi.xlsx")
nasd_onboarding = pd.read_excel(r"C:\Users\yudhaiv\OneDrive - Manulife\Project\Contact Agent Active\NASD Onboarding 05102026.xlsx")
email_ga = pd.read_excel(r"C:\Users\yudhaiv\OneDrive - Manulife\Project\Contact Agent Active\NASD 2023.xlsx")
```

```python
email_ga = pd.read_excel(r"C:\Users\yudhaiv\OneDrive - Manulife\Project\Contact Agent Active\Master Email GA.xlsx")
```

```python
print(nasd.columns.tolist())
print(nasd_onboarding.columns.tolist())
print(email_ga.columns.tolist())
```

```python
nasd = nasd[nasd["TRMN_DT"].isna()]
```

```python
nasd["Agent_NEW_RECRUIT_DATE"] = pd.to_datetime(nasd["Agent_NEW_RECRUIT_DATE"],errors="coerce")
nasd = nasd[nasd["Agent_NEW_RECRUIT_DATE"].notna()]
```

```python
nasd =  nasd[
    ~nasd["AGT_ORI"]
    .astype(str)
    .str.startswith("99B")
]
```

```python
nasd =  nasd[
    ~nasd["AGT_NM"]
    .astype(str)
    .str.upper()
    .str.startswith("ACT ")
]
```

```python
# Buat file final dulu dari NASD
final_df = nasd[
    [
        "AGT_ORI",
        "AGT_CODE_HYB",
        "AGT_NM",
        "MOBILE_PHONE",
        "OTHER_EMAIL",
        "EMAIL_ADDR",
        "BR_CODE",
        "DISTRICT_NM",
        "DIVISI_NM",
        "RANK_CD_ORI",
        "CHANNEL",
        "REGION_NAME",
        "RCAO",
        "AD_TH_RAH",
        "BDM_NAME",
        "RS_COORD",
        "RS_NAME",
        "MANUPRO",
        "TRAIN_COORD",
        "TRAINER",
        "BUILDER",
        "Agent_NEW_RECRUIT_DATE",
        "ACTIVE_SELLING",
        "HYBRID_IND"
    ]
].copy()
```

```python
final_df["BANCH_NM"] = ""
final_df["EMAIL_GA"] = ""
final_df["CITY_OFFICE"] = ""
```

```python
print(final_df.columns.tolist())
```

```python
city_mapping = nasd_onboarding.set_index("AGT_ORI")["CITY_OFFICE"]

final_df["CITY_OFFICE"] = final_df["AGT_ORI"].map(city_mapping)
```

```python
email_ga_mapping = email_ga.set_index("DIVISI_NM")["EMAIL_GA"]

final_df["EMAIL_GA"] = final_df["DIVISI_NM"].map(email_ga_mapping)
```

```python
final_df.to_excel(r"C:\Users\yudhaiv\OneDrive - Manulife\Project\Contact Agent Active\Agent_Active.xlsx",index=False)
```
